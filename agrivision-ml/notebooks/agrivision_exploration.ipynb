This notebook demonstrates exploratory data analysis (EDA), price volatility across Karnataka APMC mandis, arrival elasticity, and model inference for Tomato (*Solanum lycopersicum*).

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure src is discoverable
sys.path.append(os.path.abspath('..'))
from src.feature_engineering import create_time_features

data_path = '../data/processed/clean_tomato_data.csv'
df = pd.read_csv(data_path)
df['date'] = pd.to_datetime(df['date'])
print(f"Loaded {len(df)} APMC records across {df['market'].nunique()} markets.")
df.head()
```

```python
# Price Distribution across Mandis
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='market', y='modal_price', palette='Set2')
plt.title('Tomato Modal Price Distribution Across Karnataka Mandis (₹/Quintal)', fontsize=12)
plt.xlabel('APMC Market')
plt.ylabel('Modal Price (₹/Quintal)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Arrival vs Modal Price Scatter (Supply Elasticity Proxy)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='arrivals_qtl', y='modal_price', hue='market', alpha=0.5)
plt.title('Wholesale Arrivals vs Modal Price (Mandi Supply Elasticity)', fontsize=12)
plt.xlabel('Arrivals (Quintals)')
plt.ylabel('Modal Price (₹/Quintal)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
